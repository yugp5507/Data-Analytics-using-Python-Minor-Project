```python
full_range = set(range(1001, 10043))
existing = set(arr)
missing = sorted(full_range - existing)
print(f"Missing values count: {len(missing)}")
print(missing)
```

```python
import requests

results = []
for i in missing:
    url = f"https://ums.vnsgu.net/Result/StudentResultDisplay.aspx?HtmlURL=7596,{i}"
    response = requests.get(url)
    # Check if data exists (non-empty content and successful response)
    has_data = (response.status_code == 200 and 
                len(response.text.strip()) > 0 and 
                "No Record Found" not in response.text and
                "Result is Not Publish" not in response.text)
    results.append({'seat_number': i, 'has_data': has_data})
    print(f"Fetched seat number: {i} - Data found: {has_data}")
```
